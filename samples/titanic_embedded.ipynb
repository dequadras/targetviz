# Titanic — TargetViz Embedded Report

Run the targetviz report for the Titanic dataset in **embedded mode** (inline in the notebook).

```python
import pandas as pd
import targetviz
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)
print(f"Loaded {titanic.shape[0]} rows, {titanic.shape[1]} columns")
titanic.head()
```

```python
# Generate the report in embedded mode (output_dir=None → no file written)
report = targetviz.targetviz_report(
    titanic,
    target="Survived",
    output_dir=None,
    pct_outliers=0.05,
)
report
```
